### Task 1: Replace Hardcoded Prompts

```python
def explain_topic(topic):
   return f"Explain {topic} in simple terms for beginners"
```

**Task:**\
Convert this into a LangChain PromptTemplate

**Requirements:**\
- No f-strings
- Use PromptTemplate
- Must be reusable


````python
!pip install langchain
````

````python
from langchain_core.prompts import PromptTemplate
````

````python
# Using the PromptTemplate
template = "Explain {topic} in simple terms for beginners"

prompt = PromptTemplate(
    input_variables=["topic"],
    template=template
)
````

````python
final_prompt = prompt.format(
    topic="Artificial Intelligence"
)

print(f"Final Prompt will look like:\n{final_prompt}")
````

````python
# Install the Dependency:
!pip install langchain-groq
````

````python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model_name="llama-3.1-8b-instant",
    temperature=0.2
)
````

````python
response = llm.invoke(final_prompt)
print(response.content)
````

## Task 4: ChatPromptTemplate System

Build a chat-based prompt system using roles:
- System → defines behavior
- User → dynamic input

**Roles:**
- teacher
- interviewer
- motivator

**Example:**
```bash
role = 'teacher'
topic = 'Neural Networks'
```


````python
from langchain_core.prompts import ChatPromptTemplate
````

````python
chat_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a professional {role}"),
    ("user", "Explain me the {topic} in simple words")
])

messages = chat_prompt.format_messages(
    role="teacher",
    topic="Neural Networks"
)
````

````python
response = llm.invoke(messages)
print(response.content)
````

````python
chat_prompt1 = ChatPromptTemplate.from_messages([
    ("system", "You are a technical {role}"),
    ("user", "Ask me 5 medium level questions on the {topic}")
])

messages1 = chat_prompt1.format_messages(
    role="interviewer",
    topic="Neural Networks"
)

response = llm.invoke(messages1)
print(response.content)
````

## Task 5: Input Validation Layer
**Create function:**
validate_inputs(audience, tone)

**Rules:**
- audience ∈ `[beginner, intermediate, expert]`
- tone ∈ `[formal, casual, fun]`

**If invalid:**
- Raise error OR
- Assign default values


````python
def validate_inputs(audience: str, tone: str):
    audience_list = {'beginner', 'intermediate', 'expert'}
    tone_list = {'formal', 'casual', 'fun'}

    if audience.lower() not in audience_list:
        raise ValueError(f"The given audience-{audience} is not in audience list")

    if tone.lower() not in tone_list:
        raise ValueError(f"The given tone-{tone} is not in tone list")

    print("The values are correct")
````

````python
validate_inputs("beginner", "fun")
````

````python
validate_inputs("professional", "fun")
````

````python
validate_inputs("expert", "funny")
````

## Task 6: Prompt Generator App
**Create function:**\
`generate_prompt(topic, audience, tone, style)`


**Requirements:**
- Use PromptTemplate
- Styles: teaching, interview, storytelling

**Example Output:**\
Explain Neural Networks for beginners in a fun storytelling style


````python
def generate_prompt(topic: str, audience: str, tone: str, style: str):
    template = "Explain {topic} for {audience} in a {tone} {style} style"

    prompt = PromptTemplate(
        input_variables=["topic", "audience", "tone", "style"],
        template=template
    )

    return prompt.format(
        topic=topic,
        audience=audience,
        tone=tone,
        style=style
    )
````

````python
generate_prompt("Neural Network", "beginners", "formal", "teaching")
````

````python
generate_prompt("Neural Network", "expert", "formal", "interview")
````
